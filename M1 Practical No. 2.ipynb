```python
import pandas as pd
df = pd.read_csv("Data Set- Inc5000 Company List_2014.csv")
print(df.head())
```

```python
# Select numerical column
revenue = df["revenue"]
```

```python
print("Mean :", revenue.mean())
```

```python
print("Median :", revenue.median())
```

```python
print("Mode :")
print(revenue.mode())
```

```python
print("Minimum :", revenue.min())
```

```python
print("Maximum :", revenue.max())
```

```python
print("Range :", revenue.max() - revenue.min())
```

```python
print("Variance :", revenue.var())
```

```python
print("Standard Deviation :", revenue.std())
```

```python
print("Count :", revenue.count())
```

```python
print("Sum :", revenue.sum())
```

```python
print("\nDescriptive Statistics")
print(revenue.describe())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(8,3))
sns.boxplot(x=revenue, color="orange")
plt.title("Box Plot of Revenue")
plt.xlabel("Revenue")
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,5))
plt.hist(revenue, bins=10, color="skyblue", edgecolor="black")
plt.title("Histogram of Revenue")
plt.xlabel("Revenue")
plt.ylabel("Frequency")
plt.show()
```
